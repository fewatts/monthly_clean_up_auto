```python
import json

# --- FILE PATH SETUP ---
mosaiq_hierarchy = r'C:\Users\falve11887\OneDrive - Elekta\Documents\Monthly Clean up automations\products_hierarchy\MOSAIQ.json'

# --- LOAD HIERARCHY DATA ---
# Load the complete JSON hierarchy structure directly without modifications
with open(mosaiq_hierarchy, 'r', encoding='utf-8') as file:
    mosaiq_hierarchy_dict = json.load(file)

# --- RECURSIVE EXTRACTION FUNCTION ---
def extract_all_product_codes(nodes):
    """
    Recursively traverses the entire JSON structure to collect 
    all unique 'Product Code' values found across all tree branches.
    """
    extracted_codes = []
    # Ensure the input node/nodes structure is treated as a list
    node_list = nodes if isinstance(nodes, list) else [nodes]
    
    for node in node_list:
        if isinstance(node, dict):
            if "Product Code" in node:
                extracted_codes.append(node["Product Code"])
            # If child nodes exist, dive deeper down the tree branches
            if "Children" in node:
                extracted_codes.extend(extract_all_product_codes(node["Children"]))
    
    # Use set() to eliminate any duplicated codes from the final list
    return list(set(extracted_codes))

# --- EXECUTE EXTRACTION ---
# Get all unique raw codes directly from the complete hierarchy dictionary
final_extracted_codes = extract_all_product_codes(mosaiq_hierarchy_dict)

# --- CHUNK PACKING ALGORITHM ---
# Set the maximum character length threshold allowed per report block
CHARACTER_LIMIT = 1240 

final_report_blocks = []
current_working_block = []

for code in final_extracted_codes:
    # Calculate what the total character length would be if this code was appended
    potential_length = len(",".join(current_working_block + [code]))
    
    if potential_length <= CHARACTER_LIMIT:
        current_working_block.append(code)
    else:
        # If adding the code breaches the threshold, pack the current block and reset
        final_report_blocks.append(",".join(current_working_block))
        current_working_block = [code]

# Append the final remaining codes block if it contains data
if current_working_block:
    final_report_blocks.append(",".join(current_working_block))

# --- PRINT REPORT BLOCKS ---
# Output each individual text block with its verified character metadata
for index, block in enumerate(final_report_blocks, 1):
    print(f"--- BLOCK {index} ({len(block)} characters) ---")
    print(block)
    print("\n")

# ==============================================================================
# --- COMPLETE FULL LIST OUTPUT (FORMATTED WITH SINGLE QUOTES) ---
# ==============================================================================
# Format each code with single quotes: 'code1', 'code2', 'code3'
full_single_string_quoted = ", ".join([f"'{code}'" for code in final_extracted_codes])

print("=" * 80)
print(f"--- COMPLETE FULL LIST ({len(final_extracted_codes)} items | {len(full_single_string_quoted)} characters) ---")
print("=" * 80)
print(full_single_string_quoted)
```

```python
import os

def encontrar_prefixos_otimizados(lista_codigos, min_len=5):
    codigos = sorted(list(set(str(c).strip() for c in lista_codigos if str(c).strip())))
    if not codigos: return []
    prefixos_finais, i = [], 0
    while i < len(codigos):
        atual, j = codigos[i], i + 1
        prefixo_comum = atual
        while j < len(codigos):
            prox = codigos[j]
            common = os.path.commonprefix([prefixo_comum, prox])
            if len(common) < min_len: break
            prefixo_comum = common
            j += 1
        if len(prefixo_comum) >= min_len:
            prefixos_finais.append(prefixo_comum)
            i = j
        else:
            prefixos_finais.append(atual)
            i += 1
    return prefixos_finais

def quebrar_em_blocos(lista_itens, limite_chars=1240, separador=", "):
    blocos, bloco_atual, tamanho_atual = [], [], 0
    for item in lista_itens:
        item_str = str(item).strip()
        tamanho_item = len(item_str) if not bloco_atual else len(item_str) + len(separador)
        if tamanho_atual + tamanho_item > limite_chars:
            blocos.append(separador.join(bloco_atual))
            bloco_atual = [item_str]
            tamanho_atual = len(item_str)
        else:
            bloco_atual.append(item_str)
            tamanho_atual += tamanho_item
    if bloco_atual: blocos.append(separador.join(bloco_atual))
    return blocos

all_pc_sw = ['45010004000IQRO', '45011001201IQ', '55500001000IQKT', '45014022000IQ', '60402000KT0RO', 'CHIMPORT', '45010003302IQRO', '55101001001IQRO', '8002545', '80500000000RO', '62102000000MO', '60201244000RO', '40500001001RO', '40100003000RO', '10001000002RO', '55000000000IQ', 'TPP-PALABRAMINI', '60503003001RO', '15400000000IQMO', '45010003201IQRO', '60401000KT0MO', '63101151440RO', '#TRIALS_RO', '90032000000MO', 'TPPLSRDOCPRINTC', '45012010102IQRO', '45012003101IQ', '#BILLING_RO', 'MPS00000018', '60701000KT0IQRO', '63101151580MO', '50502011000RO', '55102000000IQRO', '52300000000IQRO', '20000000001MO', '60401255KT0MO', '63001151142RO', '63101151410RO', 'MPS00000015S', 'TPPESIPPFB', '45011000000IQRO', '97300000001MO', '45010001410IQRO', '45014022003IQRO', '45013000000IQRO', '63101151340RO', '40001005000RO', '10000000000RO', '63101151000MO', '40400020000RO', '55500009000IQRO', '63001151190MO', '45010001209IQRO', '15800000039', '90000000206', '45014024001IQRO', '63001151260RO', 'TPPESIPPCS', '10002NA0000MO', '15250000000IQ', '63101151520RO', '45011001000IQ', '60102000000MO', 'C#89513-10', '15000000000IQ', 'TPP-PALTELS-10+', '45010003300IQRO', '63001151560MO', '55105000000IQRO', '60201255KT0MO', 'MRT10131', '40000001101RO', '80100000000RO', '63001151420RO', '20200000000RO', '45001000000IQRO', '15800000005', 'PCK-EMP-RO', '45014004000IQRO', '10901000000RO', '62204000OTDI', '45010001203IQRO', '40500001000IQRO', '62701000OTDI', '25001000000IQMO', '15800000033', '99600000000', 'MRT13311', '90041000002RO', '63001151120RO', '90200000000TR', '81000000001MO', '50501001002RO', '45010003200IQRO', '15510000000IQMO', '55400001440IQRO', '55101022004IQRO', '#SMARTCLINIC_SW', '60001246000RO', '60710000000RO', '45010001400IQ', 'TPPLSR-SYQPC', 'MPS00000017S', '90000000208', '45010027001IQRO', '60603000KT0IQRO', '15800000023', '62803000OTDI', '40003003000IQRO', 'TPP-EQAPP-INT01', 'TPPLANKR672UT', '90000000216', 'TPPESIPPAR', 'PCK-ESS-RO', '90000000207', '97605000000DS', '40000001102RO', '55102011002IQRO', '63001151140MO', '45010001201IQ', '40500023001RO', '99100000001MO', '60102249000RO', 'MRT10881', '60102248000RO', '15500000000IQRO', '45010005203IQRO', '90100000002RO', '60001004000MO', '63101151400MO', '40000003205RO', '10300000000RO', '50100000000IQRO', '62108000000RO', 'PCK-ELE-RO', '45010001306IQRO', '54100003000IQRO', '60601000KT1', '99500000001RO', 'TPPLANRB422UT', '45014001003IQ', '45010003100IQRO', '97402000001DS', '99100100000MO', '45010001102IQRO', '90401607110RO', '63101151150MO', '55105001001IQKT', 'FHIR-007', '45010001000IQ', '45010000000IQRO', '#SMARTCLINIC', 'MRT10061', '60601000KT0RO', 'MPS00000019', '10400000000RO', '62304000OTDI', 'MRT10051', '55105001001IQRO', '45010002204IQRO', '60001247KT0RO', '15520000000IQRO', '15510000000IQRO', '60102244000MO', '45010003102IQRO', 'TPPESIPPPR', '62109000000RO', '76000000001DS', '55201000000IQ', '45010004101IQRO', '62109000KT0RO', '62106000KT0RO', '40001001101RO', '60712000000MO', '20000000001IQRO', '31001000KT0MO', '15400000000IQ', 'TPPESIPPCF', '40000001200RO', 'TPP-SMRTCLN-MO', '45014023001IQRO', 'MPS00000004', '97610000000DS', 'MPS00000005S', '98007000000IQMO', '20100000001MO', '15800000016', '62103000000MO', '15800000011', '63101151120MO', '55103000000IQ', '63101151510MO', '45010003208IQRO', '63101151140MO', '15800000008S', '60001240000MO', 'TPPLSRSEQPC', '10001000001MO', '63101151230MO', '40001005102RO', '55105002001IQKT', 'OS-24X7-SUPP', '20500000000RO', '40500001000RO', '45014020001IQ', '40400020001RO', '40000003202RO', '90300000011', '62401000OTDI', '60801000000MO', 'TPP-DMSKLICIRE', '40000003302RO', '55104001001IQRO', '#MOSAIQASSIST', '45012001000IQRO', '63101151590MO', '62103000000RO', '45010002207IQRO', '60001000KT0MO', '98003100000IQMO', '99100200000MO', '63000000000RO', '40100004000IQ', '90100000003MO', '50501000000RO', '50300014006RO', '45016001102IQRO', '45014021001IQRO', '99300000001RO', 'MOSAIQ-30', 'LM09999835', '90000000202', '30000000KT2MO', '90000000210', '63101151142MO', '53200000000IQMO', '90300000013', '62109000KT0MO', '63001151110MO', '15400000000IQRO', '60102245000MO', '610000000000RO', '45000000000IQKT', 'L:REPORTER', '90041000000MO', '#MOSAIQ_MO', '50502000000RO', '62107000KT0RO', 'L:PND', '#IQ_SCRIPTS_MO', '40500020001RO', '45010004103IQRO', '60701000000IQRO', '55400003000IQRO', '80200000000MO', '99200000001RO', '97551300000IQMO', '60401000000RO', '55106001000IQRO', '45010001207IQRO', 'MPS00000014S', '60001000KT0RO', '62103000KT0RO', '55101001002IQRO', 'TPPC9928A', '63100000000MO', '42500000000RO', '40500021001RO', '45010004100IQ', '63001151560RO', '45010001206IQ', '45010001302IQRO', '97551400000IQRO', '45010001100IQ', '55500008000IQKT', '50508000001RO', '10800000000MO', '80300000000MO', '40500004000RO', '40200010102RO', 'LM10568446', '76100000000DSS', '15400000KT1IQRO', '60801000001RO', 'OS-WKEND-UG-ONS', '63101151142RO', '15800000035S', '55101002000IQ', '60502000KT0', '45010001107IQ', 'LM10568441', 'TPPLSRIGRTPCE', '15800000002', 'MPS00000021S', '15800000024', 'TPPESIPPBB', '40000001206RO', '63001151440MO', '97551500000IQMO', '40000003303RO', '63001151440RO', '40001001204RO', '90511100000IQMO', '45010001107IQRO', '15800000031S', '63001151270MO', '45010028001IQRO', 'OS-TST-SRV-MAIN', '45011005102IQRO', '76000000007DS', '45011001101IQ', '99100200000RO', 'TPPLAN339317001', '55105003002IQRO', '55201000000IQRO', '31002000KT0MO', '63101151190MO', '45012001101IQ', 'MPS00000028', '55500001440IQRO', '83000000013', '45014025000IQRO', 'C#89512-10', '50507000000RO', '40000002202RO', '90000000219', '62105000KT0RO', '90100000043RO', '45720000000IQRO', '90201000000MO', '45013000000IQ', '55200000000IQ', '60201000000RO', '45011001204IQRO', '82000000000MO', '40500022001RO', '#SMARTFLOW_SW', '#MOSAIQ', '60102245000RO', '40000003201RO', '81000000004MO', '#MOSAIQ_RO', '62104000000MO', '97551500000IQRO', '63001151190RO', 'TPP-DMSKLICITAL', '97300000002RO', '55500003000IQRO', '76000000002DS', '45010001102IQ', '45010002106IQRO', '63101151480RO', '15800000004', 'C#88600', '55101022002IQ', '55500007000IQRO', 'TPPLAN225075001', '15800000017S', '45011002201IQRO', '45014022000IQRO', '45012010102IQ', '50300014007RO', '15500000000IQMO', '30000000KT0MO', '50300000000RO', '45010003103IQRO', '55400001000IQ', '97300000003MO', '99301000001RO', '62108000000MO', '83000000010', '10400000000MO', '45014-022001-IR', '55101022000IQ', '50303000000IQRO', '60201251KT0RO', '31002000KT1MO', '40001002000RO', '45012002202IQRO', 'TPPSYPMAPVER3', '97552000106', '51202000000RO', 'TPPESIPPFBI12', '76000000003DS-D', '40500030001IQRO', '40000001109RO', '62110000KT0RO', '55102011000IQ', '45010004102IQRO', 'TPP034', '63101151560RO', 'L:NDICOM DFI', '50501001001RO', '45011001301IQRO', '52000000000IQRO', '90202000000TR', '53100000000IQMO', '25500000000IQ', '15250000000IQMO', 'MPS00000003', '15300000000IQRO', '62109000000MO', '98009000000IQMO', '45011001301IQ', '81000000000MO', 'MRT10181', '80200000000RO', '45010001420IQRO', '40000001204RO', 'MRT10101', '15400000KT1IQMO', '55410000000IQRO', 'MRT10111', '26000000000IQRO', 'TPP000LISPC3', '55101022000IQRO', '53100000000IQRO', '25000000000IQMO', '63101151240MO', '60001245000RO', '90000000214', '83000000002MO', '60302000000RO', '50401000000RO', 'MPS00000020', 'SS76000000004-Z', 'TPP45 09 431', '45015000003IQRO', '81000000000RO', '99201000001MO', '45010002100IQRO', 'OS-WKEND-UG-REM', '55101001000IQRO', '63101151470MO', '40200002000RO', '46100-003000-IQ', '4514020010IQRO', '55104001001IQ', 'TPPC9932A#A2L', 'MRT10121', '63001151510MO', '42500001000RO', 'TPP-DMSKLICSPAI', '40001005101RO', '40000003105RO', '40001001301RO', '45012000000IQRO', '90000000204', 'TPPLAN2215291', '45010001101IQ', '40000003103RO', 'TPP-DMSKLICNZ', '64505000000MO', 'MPS00000003S', '45010003207IQ', '81000000003MO', '90200000005TRRO', '40001000000RO', 'LM10568436', '62503000OTDI', '45010004100IQRO', '15800000036S', '62802000OTDI', '81000000002RO', '50500000000RO', '45014020001IQRO', '15800000026', '40400025001RO', '62103000KT0MO', '45011001204IQ', '45010001101IQRO', '63101151120RO', '90401608010RO', '45010002104IQ', 'MOSAIQ-26', '45010003205IQRO', '45011002000IQRO', 'MPS00000001S', '76000000005DS', '90203000003TRRO', '40001001200RO', 'TPPLADICBLB', '40001003102RO', '60102240000MO', '45010001208IQRO', '20100000000MO', '55104000000IQ', '45011003102IQRO', '40500001002RO', '63101151250MO', '40000003203RO', '83000000001MO', 'MPS00000022S', 'SS76000000005-Z', '40000001106RO', '15300000000IQMO', '45010002105IQRO', '10700000000RO', '63001151120MO', '31004KT0000', 'MPS00000004S', '60201251000RO', '11003000000MO', '45011005101IQRO', '10300000000MO', '62804000OTDI', '40500020002RO', 'SEQ IMRT TOTAL', '45011002101IQRO', '97300000002MO', '45010002205IQRO', '90203000002TR', '45010002000IQRO', '60201000LIS', 'TPP-DMSKLICFRAN', '97551400000IQMO', '45014025001IQRO', '15200000000IQMO', '45010001106IQRO', '54001003000IQRO', '90200000004TRRO', '63101151110RO', 'MOSAIQ-27', '15800000001', '63001151000MO', '60501000000RO', 'L5854612', 'MRT10071', '80200000000', 'FHIR-002', '63101151400RO', '45011007101IQRO', '63101151390MO', '90000000201', '40500020000RO', '20000000000RO', '45010002102IQRO', 'MPS00000002S', '25000000100IQ', '63001151400MO', '#SMARTFLOW', '40000001201RO', '40000003200RO', '60102000KT0MO', '60402000KT0MO', 'TPP-PALABRA+SUB', '#MOA', '25510100000IQMO', '99500000001MO', '50501010001RO', '97552000105', 'TPPLANRG575AWOL', '45011002101IQ', '15800000025', 'TPP-EQAPP-SW', '45012010000IQRO', '45014022002IQ', '50401002001RO', '26000000000IQMO', '45010001201IQRO', '60001248000RO', '62704000OTDI', '45010003301IQRO', '63001151142MO', '45014001000IQ', '45010002103IQ', '60102000000RO', 'TPPMDDDBSRV', 'FHIR-001', '45011003000IQRO', '55499000000IQRO', '90000000213', '10900000000RO', '60001000000MO', '90041000004RO', 'LM10568438', '60302244000RO', '15800000012', '60801000001MO', 'TPP-DMSPKITCA', '20400000000MO', '#MOSAIQVOICE', '45700000000IQRO', '45011001100IQRO', '#EQUICARE_SW', '63101151140RO', '60712000000RO', '20501000KT0MO', '50201000000RO', '10002NA0000RO', '99202000001MO', '40000001205RO', '55105000000IQKT', '45010004103IQ', '40001003000IQRO', '40200000000RO', '40001007000RO', 'MPS00000018S', '45010002300IQRO', '15530000000IQRO', '76100000000DS', '55500001430IQRO', '55101004001IQ', '40000003101RO', '45014022002IQRO', '45011001440IQRO', '40000001203RO', '63101151520MO', 'MRT10861', '45014000000IQRO', 'L:BARCODE', '60001247000RO', '62105000000RO', 'TPP2824212001', '20500000000MO', '42500003000RO', '55105002001IQ', '60501000KT0', '40000003301RO', '25520000000IQMO', '45010001301IQRO', 'SS25000000000-A', '40500003000RO', '45010001400IQRO', '15250000000IQRO', '55102120001IQ', '60102246000MO', '62302000OTDI', 'TPPE0XXX', '40000001105RO', '15400000001IQMO', '10000KT0103RO', 'TPPLAN2223853', '15260000000IQMO', '30000000000MO', '62801000OTDI', '63101151220MO', '40500025000RO', '50506000000RO', '45014022001IQRO', 'MRT10191', '45011001401IQRO', '15800000035', '45015000005IQRO', '63101151310MO', '60302000KT0', '#MOSAIQASIST_SW', '98008000000IQMO', '97660000000DS', '60601000KT0', '40200002101RO', '40000001104RO', '90201000002', '62201000OTDI', 'TPPESIPPRM', '#ESI_MO', '60201000LISRO', '50503000000RO', '62102000KT0MO', '62106000000MO', '50301000000IQRO', '45014000000IQ', '45016002101IQRO', '80400000000RO', '90401607100RO', '15300-000000-IR', '10001000002MO', '40003002000RO', '40500000000RO', '46100003020IQRO', 'TPP-DMSKLICUK', '25000000000IQKT', '#IQ_SCRIPTS_RO', '60402000KT0', '45014004001IQ', 'TPP282421100001', '45701000000IQRO', '45730000000IQRO', 'TPP000SERVER1', '15600000000IQRO', 'TPPMDDQRSRV', '45011002000IQ', '60103252KT0', 'TPPLAN440189001', '90300000012', '55101010001IQ', '45011001304IQRO', '40000003102RO', '90000000215', 'L:NDICOM EPID', 'TPP000LISPC2', '45010001208IQ', '25500000000IQMO', '76403000000DS', '53400000000IQRO', '63001151370MO', '63100000000RO', '76000000008', '97551300000IQRO', 'TPPL1910A#B1H', 'C#89511-12', 'TPP-DMSKLICAU', '15800000037', '63101151190RO', '15200000000IQ', '90203000001TR', '63101151370MO', '60401000KT0', '45010001108IQRO', '62501000OTDI', '11000000000MO', '40000003207RO', '45010-003207-IR', '55101004000IQ', '40502000000RO', 'TPP-PALDMSPKITS', '83000000011', '52200000000IQRO', '45014005110IQRO', '50507001001RO', 'MPS00000016S', 'TPPESIPPSM', '45015000006IQRO', '50502011002RO', '50501022000RO', '45010002107IQ', '15000000000IQMO', 'TPP-EQAPP-SUB02', '40001002101RO', '45010001304IQRO', '90200000006TRRO', '45010002206IQRO', '90600000004MO', '50300014004RO', '31004000000MO', '55300000000IQ', 'L:DCM IMPRTP', '90000000203', '15800000003', '20400000000RO', '60101244KT0RO', '40000002200RO', '63101151360MO', '60801000KTOMO', '40501000000RO', '8002546', '99204000001MO', '53200000000IQRO', 'TPPLAN374489001', '40000001108RO', '50220000000RO', '99207000001MO', '15800000019S', '81000000003RO', '20200000000IQRO', 'TPPLSRPACSQRSRV', '45010001106IQ', 'TPP-PALABRATELS', '45010004102IQ', '90000000217', '15400000001IQRO', '45014027001IQRO', '60101000000RO', '62110000000RO', '30002000000MO', '45010001205IQRO', '60001244000MO', '55104000000IQRO', '40500022002RO', '40100001000RO', '40000001210RO', 'FHIR-008', '25000000000IQRO', 'MPS00000009', '45014020003IQRO', '20100000000RO', '63101151560MO', '45010001207IQ', '63001151500RO', '99206000001MO', '60001244000RO', '01000000000RO', '50501004001RO', '50300014003RO', 'TPPLSRDOCSCANNE', 'MOSAIQ-264', '98002100000IQMO', '76400000000DS', 'MRT10871', '60101000LIS', '97300000001RO', '50507003000RO', '#MOA SW', 'TPPMDDIMGSRV', '15520000000IQMO', '10901000000MO', '45016001101IQRO', '99200000001MO', 'TPP000SERVER2', '60001243000MO', '45014022004IQRO', '#SYNERGISTIQ', 'TPPLAN311143001', '90000200000MO', '15800000015', '60001241000RO', 'MOSAIQ-31', '15530000000IQMO', '60102247KT0', '10002NA0000', '50508000003RO', '50501004000RO', '40200010101RO', '15800000029', 'MRT15831', '76000000004DS', '45014001002IQ', '45011001400IQRO', 'SS76000000002-Z', '45011006000IQRO', '40003001000RO', '63001151410RO', '62106000000RO', '40200001000RO', '63001151450MO', 'MRT13321', '55300000000IQRO', '45010001202IQ', '62101000000MO', 'TPP-PALTELS-1-4', '50501002000RO', '45010001103IQ', '90000000209', '50401002000RO', '62107000000RO', '55105003001IQRO', '63001151240MO', 'TPP022', '55510003001IQRO', '45010001108IQ', '62102000000RO', '45011007000IQRO', '15800000030S', '60102247KT0RO', '40500000000IQRO', '99707000001RO', '50600000000RO', '50507001000RO', '82000000000RO', '45014001002IQRO', '99501000001RO', 'MRT10201', '63001151520MO', '60001246000MO', '46100003000IQ', '50507003002RO', '40001003101RO', '62604000OTDI', '60701000KTOIQRO', '#MEDICAL_ONC', '45010002208IQRO', '91200000000INRO', '15800000002S', '90000000218', '62602000OTDI', '50401001001RO', '50507002001RO', '60302000KT0RO', '80400000000', '62702000OTDI', '10000000000MO', '40400022001RO', '97552000108', '52100000000IQ', '98001100000IQMO', '83000000016', '45001000000IQ', '45010002206IQ', '15200000000IQRO', '45011001405IQRO', 'MPS00000017', '20000000001RO', '60102241000MO', '40002000000RO', 'TPP-EQAPP-SUB01', '25540100000IQMO', '15800000018S', '76000000006DS', '63001151175MO', '98010000000IQMO', 'TPPSYPVER3', '60102241000RO', '63001151150MO', '45000000000IQRO', '45010003206IQRO', '40000003100RO', '40500010001RO', '40100002000RO', '63101151130MO', '93600000000MO', '11001000000RO', '60102247000MO', '55400001000IQRO', '90032000000RO', '60201253000RO', '20000000000UR', '15800000013', '62601000OTDI', '97552000107', '20000000000MO', '90041000000RO', '54000003000IQRO', '60711000000RO', '45010003409IQRO', '#CONNECTIVITY', '40001001100RO', '55105002000IQRO', '45010001100IQRO', '15800000014', '45010002203IQRO', '80300000000', '55100000000IQRO', '62101000KT0', '90400000000MO', '46100003000IQRO', 'OS-24X5-SUPP', '55201000000IQKT', 'OS-PRACT-RF', '60102247000RO', '15600000000IQMO', '60102004000MO', '#TRIALS_MO', '60401000KT0RO', 'L:NDICOM TOTAL', '45010003000IQRO', '60201245000RO', '40400023001RO', '55500008000IQRO', '45010001105IQ', '90401607120MO', '45010003202IQRO', '15800000018', '45010001305IQRO', '55101004001IQRO', '60312000000RO', '63001151220MO', '63101151180MO', '90041000001RO', '62101000000RO', '45010001203IQ', '63101151420RO', '62202000OTDI', '50507002000RO', '45012003000IQRO', '45010001204IQ', '40000001207RO', 'MRT15861', 'C#89512-13', '60001000000RO', '#MOSAIQ_SW', '62504000OTDI', '60502000000RO', '50401001002RO', '60001245000MO', '50501022003RO', '62101000KT0RO', '63001151310MO', '20300000000MO', 'TPPLAN20490010', '15800000017', 'TPPLAN311133001', '45010002101IQ', 'MRT16521', '60403000000RO', '90000000212', '45011001201IQRO', '83000000015', 'TPPBCODEPRINT', '63001151480RO', '40500001003RO', '31003KT3MO', 'TPP-PALTEL-5-10', '#SCHEDULING_MO', '45012002000IQRO', '40000027001RO', '45012003000IQ', 'TPPDIIMB02', '15300000000IQ', '25000000131IQ', 'TPPESIPPFBI13', '63101151220RO', '45010002100IQ', '45017010001IQRO', '45010002200IQRO', '40700000000IQRO', '25000000000IQ', '45016003101IQRO', '40200001101RO', '40000000000RO', '27000000000IQRO', '30001000000MO', '60102240000RO', '40000001107RO', '40200003101RO', '50200000000RO', '51203000000RO', '#DATA_DIRECTOR', 'MPS00000005', '45012002201IQ', 'MPS00000010S', '50506001001RO', '50200000000IQRO', '63101151420MO', '45012002201IQRO', '90600000000MO', '60701000KT0IQMO', '45012003101IQRO', '45010001205IQ', 'MPS00000015', 'FHIR-005', '45010002201IQRO', '25520100000IQMO', '45014026001IQRO', '50100000000MO', '90401000001', '62703000OTDI', '55105002000IQ', '60402000000MO', '55106003000IQ', '52400000000IQRO', '45014022001IQ', '99302000001MO', '#MSQTELEPORT', '90600000004RO', '40000001103RO', '45010001103IQRO', '50501022002RO', '63001151590MO', 'TPPESIPPFBI3', 'MRT15811', '90401000000', '15800000030', '40100000000RO', '10001000001RO', '63101151000RO', '63001151230MO', 'FHIR-003', '20200000000MO', '45500005203IQRO', 'MPS00000010', '15800000033S', '60102000KT0RO', '99302000001RO', '11001000000MO', '10700000000MO', '45011005000IQRO', '90300000014', 'MPS00000009S', 'TPPLSRSYQ1MON', '40200002202RO', '90041000003RO', '60001241000MO', 'MPS00000007S', '45010001104IQRO', '#MOSAIQ_RTP', '#EVALUATE', 'MOSAIQ-32', 'CHGXPORT', '83000000014', '#PARTICLES', 'PCK-EMP-CO', 'TPPLSRPACSDBSRV', 'TPPPASYMMAPLIC', 'L:MLC FITS', '90401000072S', '45010002104IQRO', '55102011001IQ', '63101151250RO', '90401607120RO', '55101004000IQRO', '50800000000IQRO', '55500001410IQRO', '45010004000IQ', '40500025001RO', '15800000036', 'MPS00000021', '62108000KT0RO', 'MPS00000016', '50502011001RO', '45011000000IQ', '40200010000RO', '55102000000IQ', '60602235000RO', '60403000KTORO', '45010001105IQRO', '62203000OTDI', '50507003001RO', 'TPPLAN305461001', '63001151000RO', '42500001001RO', 'MOSAIQ-286', 'TPPESIPPBBI13', '15800000032S', '15900000000IQMO', 'MPS00000022', 'MRT13041', 'TPP-PALLOCUMUSR', '45012002101IQRO', '97670000000DS', '15800000010', '45010001000IQRO', '45011006000IQ', '11000000000RO', 'MPS00000007', 'TPPLAN384Z10300', '40000004000RO', '10900000000MO', 'OS-SRV-MIG', 'MPS00000006', '63001151520RO', '45010003204IQRO', 'C#89512-14', '63101151270MO', '63101151175MO', '60102243000RO', '20301000000MO', '15800000027', '60701000000IQMO', '90300000010', '25001000000IQ', '45010003303IQRO', '98011000000IQMO', '80100000000MO', '60312000KT0RO', '55101010001IQRO', '81000000002MO', 'MRT15841', '#MOSAIQ_PS_ALON', '45010005420IQRO', 'L:PRIMEVIEW', '20501000KT0RO', '60101000KT0', '55499000001IQRO', '15200000000IQKT', '97551200000IQRO', '45010004101IQ', '62104000000RO', 'TPP-EQAPP-SUB03', '45012002000IQ', '50700012000RO', '62109OOOKT0RO', '52100000000IQRO', 'TPP-DMSKLICGERM', '45010002202IQRO', '10500000000RO', '83000000000MO', '55500001000IQRO', '45710000000IQRO', 'TPPLANXW8200', '63101151110MO', '10001000000MO', '45010002107IQRO', '62107000000MO', 'OS-WKDAY-UG-ONS', '45010001440IQRO', 'PCK-ESS-MO', '40001001000RO', '40003003000RO', '60102243KT0RO', '60001240KT0MO', '45017003101IQRO', '45010001303IQRO', '45012000000IQ', '40503000000RO', 'MPS00000002', '25002000000IQRO', 'TPP-SMRTCLN-RO', 'MPS00000020S', 'MRT13051', '80300000000RO', '50100000000RO', '60401000000MO', '40003000000RO', '90203000006TRRO', '55101022003IQRO', '45010001104IQ', '60102000KT0', 'TPPC9919A#A2L', '40000001202RO', '40400000000RO', '40000003204RO', 'PCK-ELEV-RO', '55500003000IQKT', '60101235000RO', '45010003207IQRO', '25002000000IQMO', 'TPPLAN20510010', 'TPPBCODESCAN', '55105001001IQ', '55101002000IQRO', '76401000000DS', 'PCK-ELEV-CO', '55105002001IQRO', '25001000000IQRO', '10001000000RO', 'TPPPA-DSRUNLIC', '15800000034', '60101000KT0MO', '55500001420IQRO', 'TPPLAN416561001', '54004003000IQRO', '83000000012', '90100000042RO', '63001151580MO', '62404000OTDI', 'PCK-ESS-CO', '90203000002TRRO', '62101000KT0MO', '90203000003TR', '45015000000IQ', '60001247000MO', '60402000000RO', '45014001000IQRO', '62403000OTDI', '55102120001IQRO', '20300000000RO', '50700000000RO', '63101151500RO', '45014020002IQ', '40002002000RO', '60801000000RO', '90401608010MO', '10500000000MO', '40001001102RO', 'MRT13061', '90403000001', '62110000000MO', '60001243000RO', '#BILLING_MO', '50220000000IQRO', 'TPPPADSRUNLIC', '45014020002IQRO', 'L:SUPERVISOR', '60102004000RO', '60101000KT0RO', '25510000000IQMO', '90600000000RO', 'TPPESIPPBBI12', '50201000000IQRO', '60401255KT0RO', '63001151490MO', '15800000021', '63001151110RO', '45010000000IQ', '40200003000RO', '45011001000IQRO', '60101235000MO', 'MRT10161', '45010001200IQ', '15800000028', '90400000000RO', '15800000034S', 'MPS00000001', '25000000100IQMO', '45010001200IQRO', '45010001300IQRO', '99203000001MO', '40000004103RO', '62303000OTDI', '15800000019', '55103001008IQRO', '55500002000IQRO', '45010001109IQRO', 'L:ASSESSMENTS', '98004100000IQMO', '40400024001RO', '53400000000IQMO', '63001151220RO', 'TPPLAN62411601', '45011007000IQ', '60101236000RO', '63001151180RO', '45010003203IQRO', '31010000KT0MO', '45010003101IQRO', '60101236000MO', '97680000000DS', '55101000000IQ', 'MPS00000019S', '62301000OTDI', 'TPPLANRB456UT', '45010003209IQRO', '40001001300RO', '64508000000MO', '76000000000DS', '45010002103IQRO', '64509000000MO', 'LM10568442', '40002001000RO', '60001240000RO', '63101151330RO', '40000003400RO', '63101151460RO', '40500022000RO', '#ESI_RO', '55400002000IQRO', '55101002001IQRO', '90000000205', '63101151370RO', '50300014000RO', 'TPPDIIMTSUDS1P', '50400000000RO', '45014024001IQ', '98000100000IQMO', '45011001404IQRO', '45014020010IQRO', 'C#92808', '40500004001RO', 'TPP-SMRTCLN-LIN', 'MPS00000008', '45014020000IQRO', '63001151130MO', 'C#89511-13', '40001002201RO', '45014004001IQRO', '45014001001IQRO', '60101245KT0RO', '90000000220', '76402000000DS', '60101000000MO', '15000000000IQRO', '62104000KT0RO', '40000001208RO', '15800000008', 'PCK-ELEV-MO', '55106001000IQ', 'MRT13031', '40001001304RO', '25000000100IQRO', 'TPP000SERVER3', '45011006101IQRO', '63101151180RO', '63001151380MO', '20301000000RO', '50401001000RO', '51200000000RO', '63000000000MO', 'TPPPA-SYMRUNLIC', '64506000000MO', '50401000000IQRO', '60701000KTOIQMO', '45011002201IQ', 'MPS00000030', '45010001206IQRO', '50502012001RO', '45011005103IQRO', 'SS76000000001-Z', '40000003000IQRO', '63101151490MO', '62402000OTDI', '62603000OTDI', 'TPPLANRB438UT', '99205000001MO', 'TPP-EQAPP-INT02', '50302000000IQRO', '45011001102IQRO', '51201000000RO', '45011001101IQRO', '40001001201RO', '45012001101IQRO', '45012002202IQ', '63101151510RO', '55102011001IQRO', 'MOSAIQ-28', 'MRT10141', 'FHIR-004', '15800000032', '45011000000IQKT', '62105000000MO', '40400001003RO', '60403000KT0RO', '43700000000IQRO', '15800000022', '63001151420MO', '90000000200', '45011003122IQRO', '45011001304IQ', '#RADIATION_ONC', '55101022002IQRO', 'MPS00000006S', '45012010000IQ', '40001007101RO', '99100100000RO', 'TPPESIPPBBI3', '45011003121IQRO', '60102244000RO', '45010001202IQRO', 'FHIR-006', '40000002000RO', '97300000003RO', '15000-000000-IR', '60201000KT0RO', '63001151330MO', '63101151380MO', '45014020000IQ', '25530000000IQMO', '11002000000MO', '60501000KT0RO', '15800000038', '45010002101IQRO', '#SCHEDULING_RO', '63001151470MO', '55400003000IQ', 'TPP-EQAPP-INT03', '45015000000IQRO', '60502000KT0RO', '27000000000IQMO', '11002000000RO', '25505000000IQMO', 'LM10568440', '40000003000RO', '10800000000RO', '81000000001RO', '45014001003IQRO', '#MDD_SW', 'C#89512-12', 'OS-IP-CHANGE', '98006100000IQMO', '20100000001RO', '45011001102IQ', '99100000001RO', '45010002000IQ', '63001151260MO', '25530100000IQMO', '#MSQTELEPORT_SW', '55103000000IQRO', '63101151410MO', '76000000001DSS', 'MPS00000014', '55101002001IQ', 'PCK-EMP-MO', '55105000000IQ', 'MRT10151', '45011001200IQRO', '45011003101IQRO', '45014025000IQ', '80400000000MO', '40400001000RO', '55102011000IQRO', '60312000KT0', '#MOSAIQ_DESKTOP', 'TPP000LISPC1', '97606000000DS', '50501002001RO', '62502000OTDI', '90000201000MO', '98005100000IQMO', '45010001204IQRO', '10000000000IQRO', 'PCK-ELE-RO-UNL', 'TPPLAN413929001', 'L:DCM IMPRTI', '90000000211', '40000003300RO', '63101151450MO', '8001560', 'C#89555-20', 'C#89547-20', '8001561', 'C#89551', '1550232', '1550236', '1550235', '8001555', '8001565', '8001568', '8001563', '#DOSISOFT', 'C#89542', 'C#89554-10', '8001576', 'C#89541-21', '8001570', '1550234', 'C#89548-20', 'C#89543-20', 'C#89560', '1550233', 'C#89542-10', '8001571', 'C#89545', '8001425', 'C#89550', '8001577', 'C#89565', 'C#89548', '8001575', 'C#89552-20', 'C#89547', '8001562', 'C#89549', '8001557', '8001564', 'C#89555', 'C#89550-20', 'C#89549-20', '1553405', 'C#89544', 'C#89543', '8001581', 'C#89553-20', 'C#89546', 'C#89541', 'C#89559', 'C#89528', '1567024', 'C#89526', '1554715', 'C#89534', '8005003', 'C#89530-32', '1567895', '1554712', '1541637', '1517662', '1541636', 'C#89533', '8005001', '1517663', 'C#89525-32', '1554714', '#AQUA', '8005000', 'C#89529', 'C#89532', 'C#89527', '1543771', 'C#89525-33', '1543770', '8005002', 'C#89531', '1554713', '1554711', '#KAIKU', '90403000014SK', '1570000020S', '15700000017', '1570000022S', '90401608010SK', '15700000013', '15700000026', '1570000024S', '1570000025S', '15700000002', '15700000018', '15700000006', '15700000003', '60502000000SK', '15700000005', '15700000007', '15700000004', '1570000023S', '15700000012', '15700000011', '1570000021S', '60801000001SK', '15700000014', '15700000008', '#KAIKU_SW', '15700000015', '15700000016', '15700000009', '15700000027', '15700000001', '15700000010', '15700000019', '15700000028', '90000000100SK', '60000000100SK', 'C#76552', 'C#80511', 'C#80122', 'C#89508-61', 'C#92460', 'C#92549-F1', 'C#89508-601', 'C#92555', 'C#94215-F1', 'C#80126', 'C#92564', 'C#76559', 'C#89508-50000', 'C#88559', 'C#76568', 'C#89508-30100', 'C#92560', 'C#89583-13', 'C#89508-F1', 'C#92551', 'C#89508-63', 'C#88561', 'C#88523UP', 'C#89586', 'C#94284', 'C#94244', 'C#89508-530', 'C#80123', 'C#92561', 'C#72951', 'C#80125', 'TPK-96001', 'C#94851', 'C#89583', 'C#92569', 'C#MONACO', 'C#89508-550-F1', 'C#88556', 'C#92410AD', 'C#71947', 'C#89508-20400', 'C#92443', 'C#92598', 'C#94251', 'C#71937', 'C#89588', 'C#70034', 'C#92430MN-F1', 'TPK-SAAS-001', 'C#92553', 'C#91739', 'C#94210', 'C#88558-41100', 'C#88563', 'C#92559', 'C#92556', 'C#92561-F1', 'C#89508-540', 'C#89508-600', 'C#89581', 'C#94281', 'C#92457MN', 'C#89508', 'C#88556-54000', 'C#93339', 'C#92414MN-F1', 'C#94236', 'C#92430MN', 'C#92558', 'C#94207', 'C#94250', 'C#89585', 'C#89864', 'C#88543', 'TPK-92593', 'C#76565', 'C#92405MN', 'C#89421-41100', 'C#92548-F1', 'C#88539-51000', 'C#13973', 'C#94850', 'C#89508-53000', 'C#94243', 'C#92548', 'C#89508-32000', 'C#94285', 'C#88536', 'C#76555', 'C#94249', 'C#71929', 'C#92595-EDU', 'C#92461AD', 'C#80505', 'C#89508-52000', 'C#92414MN', 'C#71943', 'C#92410MN-F1', 'C#92430AD', 'C#71944', 'C#92550', 'C#89581-13', 'C#89584', 'C#71930', 'C#92549', 'C#94215', 'C#76560', 'C#88543-F1', 'C#71928', 'C#89508-511-F1', 'C#80511-F1', 'C#88523-41100', 'C#92414AD', 'C#92410AD-F1', 'C#71940', 'C#88536UP-F1', 'C#89508-20300', 'C#76566', 'C#92595', 'C#88537-F1', 'C#88537UP', 'C#89508-551', 'C#93334', 'C#92399', 'C#76547', 'C#89508-550', 'C#94247', 'C#89508-62', 'C#89508-31000', 'C#80121', 'C#92470-F1', 'C#71927', 'C#92571', 'C#71959', 'C#92557', 'C#92470', 'C#89508-510', 'C#72105', 'C#92404MN', 'PCK-DIRACC-SAAS', 'C#94248', 'C#76551', 'C#88562', 'C#92398', 'C#92552', 'C#92396', 'C#92456', 'PCK-MONACO-SAAS', 'C#89508-520', 'PCK-PLAN-PRO', 'C#88564', 'PCK-ONE-PLAN', 'C#76558', 'C#76557', 'C#88537-54000', 'C#88543UP', 'C#71939', 'C#89508-51100', 'C#92430-F1', 'C#88543-54000', 'TPK-92592A', 'C#80124', 'C#80504', 'C#89508-511', 'C#13971', 'C#92470-E', 'C#94848', 'C#88536UP', 'C#13972', 'TPK-92594-EDU', 'C#92552-F1', 'TPK-EOP-CAPEX', 'C#88523UP-F1', 'C#94208', 'C#92405AD', 'C#92597', 'C#92430AD-F1', 'C#94246', 'C#92554', 'C#92599', 'C#71948', 'C#76556', 'C#88556-50000', 'C#94240', 'C#92457AD', 'C#89514-10', 'C#92569-F1', 'C#76561', 'C#89508-51000', 'C#89514', 'C#89582-74', 'C#71950', 'C#88537', 'C#91707', 'C#89589', 'C#71933', 'C#92456-F1', 'C#92410MN', 'C#92551-F1', 'C#88536-41100', 'C#89582', 'C#94237', 'C#94283', 'C#94207-F1', 'C#94208-F1', 'C#92461MN', 'C#76564', 'C#13974', 'C#88523', 'TPK-92591A', 'C#76554', 'C#89508-30000', 'C#94234', 'C#94245', 'C#92571-F1', 'C#94280', 'C#89580', 'C#89508-33000', 'C#71949', 'C#89580-74', 'C#80804', 'C#SIM']

prefixos = encontrar_prefixos_otimizados(all_pc_sw, min_len=5)
blocos = quebrar_em_blocos(prefixos, limite_chars=1240, separador=", ")

for i, bloco in enumerate(blocos, 1):
    print(f"--- BLOCO {i} ({len(bloco)} chars) ---")
    print(bloco)
    print()
```
